## Experiment 1: Data Analysis

### Experiment 1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive

drive.mount('/content/drive')
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/CL-NLP - Lab/Lab-1/employee_information_100.csv")

print("First 5 rows of the dataset:")
print(df.head())

df.info()
```

```python
average_salary_by_department = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)

print("\nAverage Salary by Department:")
print(average_salary_by_department)

plt.figure(figsize=(10, 6))
sns.barplot(x=average_salary_by_department.index, y=average_salary_by_department.values, palette='viridis')
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.tight_layout()
plt.show()
```

Experiment 1.2: Display the number of employees working in each department.

```python
emps_by_dept = df['Department'].value_counts().sort_values(ascending=False)

print("\nNumber of Employees by Department:")
print(emps_by_dept)
```

Experiment 1.3: Display the percentage of male and female employees using a pie chart.

```python
emps = df['Gender'].value_counts()

print(emps)
```

```python
plt.figure(figsize=(8,4))

plt.pie(emps, labels=emps.index, autopct='%1.1f%%')
plt.title("Percentage of employees by gender")

plt.show()
```

Experiment 1.4: Draw a histogram showing salary distribution.


```python
# Experiment 1.4: Draw a histogram showing salary distribution.

plt.figure(figsize=(10, 6))
sns.histplot(df['Salary'], bins=10, color='skyblue')
plt.title('Salary Distribution of Employees')
plt.xlabel('Salary')
plt.ylabel('Number of Employees')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Experience_Years', y='Salary', data=df, palette='viridis', s=50, alpha=0.8)
plt.title('Relationship between Experience and Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

Experiment 1.6: Display the top 10 employees with the highest salary.

```python

display(df.sort_values(by='Salary', ascending=False).head(10))
```

Experiment 1.7: Find the highest salary in every department.


```python
df.groupby("Department")["Salary"].max()
```

Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
avg_salary = df["Salary"].sum() / df.shape[0]

employees_above_avg_salary = df[df['Salary'] > avg_salary]
display(employees_above_avg_salary)
```

Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_exp_by_dept = df.groupby('Department')['Experience_Years'].mean().sort_values(ascending=False)
display(avg_exp_by_dept)
```

Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=10, color='purple')
plt.title('Distribution of Employee Ages')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```
