## 5.1(a) BPE Tokenization using Pretrained Model

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained('gpt2')

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text, add_special_tokens=False)

print('Tokens:')
print(tokens)

print('\nToken IDs:')
print(token_ids)
````

## 5.1(b) BPE Tokenization without Pretrained Model

````python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

tokenizer = Tokenizer(BPE(unk_token='[UNK]'))
trainer = BpeTrainer(
    vocab_size=300,
    min_frequency=2,
    special_tokens=['[UNK]', '[PAD]', '[CLS]', '[SEP]']
)

tokenizer.pre_tokenizer = Whitespace()
tokenizer.train([file_path], trainer)

output = tokenizer.encode(text)

print('Tokens:')
print(output.tokens)

print('\nToken IDs:')
print(output.ids)
````

## 5.2(a) SentencePiece Tokenization using Pretrained Model

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained('google/mt5-small')

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text, add_special_tokens=False)

print('Tokens:')
print(tokens)

print('\nToken IDs:')
print(token_ids)
````

## 5.2(b) SentencePiece Tokenization without Pretrained Model

````python
import sentencepiece as spm

model_prefix = '/content/my_sentencepiece'

spm.SentencePieceTrainer.train(
    input=file_path,
    model_prefix=model_prefix,
    vocab_size=300,
    model_type='unigram',
    character_coverage=1.0,
    hard_vocab_limit=False
)

sp = spm.SentencePieceProcessor(model_file=model_prefix + '.model')

tokens = sp.encode(text, out_type=str)
token_ids = sp.encode(text, out_type=int)

print('Tokens:')
print(tokens)

print('\nToken IDs:')
print(token_ids)
````

## 5.3(a) POS Tagging using spaCy

````python
import spacy

nlp = spacy.load('en_core_web_sm')

sentence = 'The young student is reading an interesting book in the library.'
doc = nlp(sentence)

seen = set()

print(f"{'Token':<15}{'POS':<10}{'Description'}")
print('-' * 50)

for token in doc:
    if token.text not in seen and not token.is_punct:
        seen.add(token.text)
        print(f'{token.text:<15}{token.pos_:<10}{spacy.explain(token.pos_)}')
````

## 5.3(b) POS Tagging using NLTK

````python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

tokens = word_tokenize(sentence)
tags = pos_tag(tokens)

description = {
    'NN': 'Noun', 'NNS': 'Plural noun', 'NNP': 'Proper noun',
    'VB': 'Verb', 'VBD': 'Past tense verb', 'VBG': 'Gerund',
    'VBN': 'Past participle', 'VBP': 'Present verb', 'VBZ': '3rd person singular verb',
    'JJ': 'Adjective', 'JJR': 'Comparative adjective', 'JJS': 'Superlative adjective',
    'RB': 'Adverb', 'DT': 'Determiner', 'IN': 'Preposition',
    'CC': 'Conjunction', 'PRP': 'Personal pronoun', 'TO': 'Infinitive marker'
}

seen = set()
print(f"{'Token':<15}{'POS':<10}{'Description'}")
print('-' * 50)

for token, tag in tags:
    if token.isalpha() and token not in seen:
        seen.add(token)
        print(f'{token:<15}{tag:<10}{description.get(tag, "Other")}')
````

## 5.4 POS Tagging using spaCy with Frequency

````python
from collections import Counter

doc = nlp(text)

frequency = Counter(
    token.text.lower()
    for token in doc
    if token.is_alpha
)

seen = set()

print(f"{'Token':<25}{'POS':<10}{'Description':<30}{'Frequency'}")
print('-' * 80)

for token in doc:
    word = token.text.lower()
    if token.is_alpha and word not in seen:
        seen.add(word)
        print(f'{word:<25}{token.pos_:<10}{spacy.explain(token.pos_):<30}{frequency[word]}')
````

````python
!pip install -q gensim



file_path = '/content/drive/MyDrive/Colab Notebooks/CL-NLP - Lab/Lab-5/corpus.txt'

with open(file_path, 'r', encoding='utf-8') as f:
    text = f.read()

sentences = [line.lower().split() for line in text.strip().split('\n')]

# CBOW: sg=0
model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    workers=1,
    epochs=100
)

word = 'learning'

print("Vector representation of", word, ":")
print(model.wv[word])

print("\nThree most similar words:")
print(model.wv.most_similar(word, topn=3))
````

**EXP 5.6- Skip-Gram Word2Vec**

````python
from gensim.models import Word2Vec

# Skip-Gram: sg=1
model_sg = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=1,
    workers=1,
    epochs=100
)

word = 'language'

print("Vector representation of", word, ":")
print(model_sg.wv[word])

print("\nFive most similar words:")
print(model_sg.wv.most_similar(word, topn=5))
````

**EXP 5.7- Cosine Similarity**
