```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet')
try:
    nltk.data.find('taggers/averaged_perceptron_tagger')
except LookupError:
    nltk.download('averaged_perceptron_tagger')
```

### Experiment 3.1: Stemming using Porter Stemmer

```python
stemmer = PorterStemmer()
words_for_stemming = ["playing", "played", "plays", "studies", "studying", "connected", "connection", "computers"]

print("Original Word -> Stemmed Word:")
for word in words_for_stemming:
    print(f"{word} -> {stemmer.stem(word)}")
```

### Experiment 3.2: Lemmatization

```python
lemmatizer = WordNetLemmatizer()
words_for_lemmatization = [
    "cats", "dogs", "running", "runs", "ran", "studies", "studying",
    "better", "children", "mice", "went", "ate", "leaves", "caring"
]

print("Original Word -> Lemmatized Word (noun and verb forms):")
for word in words_for_lemmatization:
    lemma_v = lemmatizer.lemmatize(word, pos='v')
    lemma_n = lemmatizer.lemmatize(word, pos='n')
    if lemma_v != word and lemma_n != word and lemma_v != lemma_n:
        print(f"{word} -> Verb: {lemma_v}, Noun: {lemma_n}")
    elif lemma_v != word:
        print(f"{word} -> {lemma_v} (verb)")
    elif lemma_n != word:
        print(f"{word} -> {lemma_n} (noun)")
    else:
        print(f"{word} -> {word}")
```

### Experiment 3.3: Regular Expressions (Regex) Extraction

```python
input_text = """
Welcome to the Natural Language Processing workshop! For more information, visit https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your experience using #NLP, #Python, and #MachineLearning. You can also visit https://github.com/NLPWorkshop for the latest updates.
"""

print("Input Text:\n" + input_text)

email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, input_text)
print("\nEmail Addresses:", emails)

url_pattern = r'\b(?:https?://|www\.)\S+\b|\b\S+\.(?:com|org|net|gov|edu)\b'
urls = re.findall(url_pattern, input_text)
filtered_urls = []
for url in urls:
    if '@' not in url and not url.endswith('.'):
        filtered_urls.append(url.strip('.,'))
print("URLs:", list(set(filtered_urls)))

mobile_pattern = r'\b(?:\+91[- ]?)?[0-9]{10}\b'
mobile_numbers = re.findall(mobile_pattern, input_text)
print("Mobile Numbers:", mobile_numbers)

hashtag_pattern = r'#\w+'
hashtags = re.findall(hashtag_pattern, input_text)
print("Hashtags:", hashtags)

mention_pattern = r'@\w+'
mentions = re.findall(mention_pattern, input_text)
print("Mentions:", mentions)
```
